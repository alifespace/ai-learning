### 安装字体（适配中文与标准科技文档）
### 安装 LaTeX 编译环境
1. 安装 MiKTeX
2. 安装已下宏包
- xeCJK：中文支持
- fontspec：字体配置
- tocloft：目录美化
- hyperref：目录跳转支持
### 配置nbconvert模板
需要自定义一个`latex-chinese` 模板:
1. 目录结构建议：
```bash
.your_venv/
└── share/
    └── jupyter/
        └── nbconvert/
            └── templates/
                └── latex-chinese/
                    ├── index.tex.j2
                    └── conf.json（可选）
```
2. 示例`index.tex.j2`模板的核心内容：
```jinja2
((*- extends 'latex/base.tex.j2' -*))

((* block header *))
\usepackage{xeCJK}
\usepackage{fontspec}
\usepackage{tocloft}
\usepackage{hyperref}
\usepackage{bookmark}

\setmainfont{Times New Roman}
\setCJKmainfont{SimSun}

\title{{{{ resources.metadata.title }}}}
\author{{{{ resources.metadata.authors | join(', ') }}}}
\date{{{{ resources.metadata.date }}}}

\begin{document}
\maketitle
\tableofcontents
\newpage
((* endblock header *))
```

````python
import matplotlib.pyplot as plt
from ipywidgets import interact
import numpy as np

def plot_func(a=1):
    x = np.linspace(-10, 10, 400)
    y = a * np.sin(x)
    plt.plot(x, y)
    plt.show()

interact(plot_func, a=(0.1, 5, 0.1))

````
