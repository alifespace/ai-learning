## 配置`index.tex.j2`模板的核心内容：
主要通过配置`index.tex.j2`来完成定制化的`pdf`输出。

### Day1：实现基本的中文输出
```jinja2
((*- extends 'base.tex.j2' -*))

((* block docclass *))
\documentclass[UTF8]{ctexart}
((* endblock docclass *))

((* block packages *))
((( super() )))
\usepackage{fontspec}
\setCJKmainfont{SimSun}
\setmonofont{Consolas}
((* endblock packages *))
```
### Day2: 实现目录的能力
```Jinja2
((*- extends 'base.tex.j2' -*))

%----------------------------------------
% 中文文档类 + 支持科技论文规范
%----------------------------------------
((* block docclass *))
\documentclass[UTF8]{ctexart}
((* endblock docclass *))

((* block packages *))
((( super() )))
\usepackage{fontspec}
\setCJKmainfont{SimSun}
\setmonofont{Consolas}
((* endblock packages *))


((* block commands *))

\usepackage{xcolor}
\usepackage{etoolbox}
\usepackage{titlesec}
\usepackage{tocloft}
\usepackage{titletoc}

% 设置链接颜色黑色
\definecolor{linkcolor}{rgb}{0,0,0}
\definecolor{urlcolor}{rgb}{0,0,0}
\definecolor{citecolor}{rgb}{0,0,0}
\hypersetup{
  colorlinks=true,
  linkcolor=linkcolor,
  urlcolor=urlcolor,
  citecolor=citecolor,
}

% 清除标题染色
\makeatletter
\patchcmd{\@sect}{\color{linkcolor}}{}{}{}
\makeatother

% 一级标题样式
\titleformat{\section}
  {\normalfont\bfseries\color{black}\Large}
  {\thesection}{1em}{}
\titlespacing{\section}{0pt}{1.0ex plus .2ex minus .2ex}{0.5ex plus .1ex}

% 加粗目录条目数字
\renewcommand{\cftsecfont}{\bfseries}
\renewcommand{\cftsecpagefont}{\bfseries}
% \renewcommand{\cftsecleader}{\cftdotfill{\cftdotsep}}

% 目录标题：蓝色、大字号、不居中
\renewcommand{\contentsname}{\textcolor{blue}{\bfseries 目录}\\[1.0ex]}

% 目录标题与目录条目之间增加垂直间隙
\makeatletter
\patchcmd{\tableofcontents}{\@starttoc{toc}}{%
  \vspace{1em}\@starttoc{toc}}{}{}
\makeatother

((* endblock commands *))

%----------------------------------------
% 标题元信息
%----------------------------------------
((* block title *))
\title{ {{ resources.metadata.title | default("无标题文档") }} }
\author{ {{ resources.metadata.authors | join(", ") | default("无名作者") }} }
\date{ {{ resources.metadata.date | default("编译日期：\today") }} }
\date{{{{ resources.metadata.date | default("编译日期：\today") }}}}
((* endblock title *))

%----------------------------------------
% 启用标题 + 目录页
%----------------------------------------
((* block maketitle *))
\maketitle
\newpage
\tableofcontents
\newpage
((* endblock maketitle *))
```
