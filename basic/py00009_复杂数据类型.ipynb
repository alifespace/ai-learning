# 字符串
## 字符串概述
在`Python`中处理文本数据是使用`str`对象，也称为字符串。字符串是由`Unicode`码位构成的不可变序列。字符
串字面值有多种不同的写法。
- 单引号: ‘允许包含有“双”引号’;
- 双引号:“允许包含有‘单’引号”；
- 三重引号: 使用三重引号的字符串可以跨越多行——其中所有的空白字符都将包含在该字符串字面值中；

```python
s1 = 'Lemon'
s2 = "你好, python"
s3 = """
Hello Python World!
欢迎来到Python的世界
"""
print(s1)
print(s2)
print(s3)
```

## 字符串常见操作
字符串（string）支持用切片的方式来访问字符串中的值，这个方式与`list`数据类型的切片方式类似。

```python
s2 = "Hello Python World"
print(s2[8])
print(s2[6:])
```

打印的字符串里如果包括`\`，因为它是转义字符，所以打印这种字符串，最前面加个`r`。

```python
print(r"\data")
print(r'D:\no', "\no")
```

字符串格式化，是`Python`字符串内容的重要组成部分，应用广泛。`Python`支持格式化字符串的输出，一般有3种方式可以实现，包括`format，%，f-string`。其中`f-string`方式是在`python 3.6`之后开始支持。

```python
# format
name = "Lemon"
age = 18

print('my name is {0}, age is {1}'.format(name, age))
```

```python
# %
print("my name is %s, my age is %s" % (name, age))
```

```python
# f-string： python 3.6以上的版本，可以实现下述格式化输出的方法
print(f"my name is {name}, my age is {age}")
```

### 数字格式化
在进行字符串格式化时，经常会遇到需要经数字格式化为字符串，并且要按某种特定的格式来显示。数字格式化为字符串，可以用format、%或f-string方法来实现。下面的数字格式化，主要以`f-string`方法来举例。



```python
n1 =3.1415926
n2 =31415.926
n3 =0.31415
n4 =21

#保留两位小数的数字格式
print("保留两位小数: {:.2f}".format(n1))
print("保留两位小数: %.2f" % (n1))
print(f"保留两位小数: {n1:.2f}")
```

```python
# 百分比格式
print(f"百分比格式：{n1:.2%}")

# 既有千分位分隔符又有精度设定的数字格式
print(f"'既有千分位分隔符又有小数位数：{n2:,.2f}")
```
